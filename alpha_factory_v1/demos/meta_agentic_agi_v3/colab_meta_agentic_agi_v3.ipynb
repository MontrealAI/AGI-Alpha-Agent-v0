```python
from alpha_factory_v1.demos.meta_agentic_agi_v3.curriculum_lab import read_json, evaluate, verify, write_bundle, brief
cases = read_json(root / "alpha_factory_v1/demos/meta_agentic_agi_v3/scenarios.json")
report = evaluate(cases[0])
print(brief(report))
```

```python
for row in report["result"]["history"]:
    print("Round", row["round"], "difficulty", row["difficulty"], "selected", row["winner"], "solve rate", row["recent"]["accuracyBps"] / 100, "%")
```

```python
target = write_bundle(report, Path("curriculum-runs"))
assert verify(read_json(target / "run.json")) == report
print("Verified all evidence:", target.resolve())
```
